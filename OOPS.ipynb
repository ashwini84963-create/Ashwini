```python
class Animal:   # Method overidding
  def sound(self):
    pass

class Dog(Animal):
  def sound(self):
    print("Baww")

class Cat(Animal):
  def sound(self):
    print("Meow")

class Cow(Animal):
  def sound(self):
    print("Moo!")

animals = [Dog(), Cat(), Cow()]

for animal in animals:
  animal.sound()

dog = Dog()
cat = Cat()
cow = Cow()

dog.sound()
cat.sound()
cow.sound()
```

Abstraction


   it becomes blueprint for other classes

```python
from abc import ABC , abstractmethod     # import tools for abstract classes

class Animal(ABC):    # Create an abstract parent class

  @abstractmethod   # Reqiure child classes to implement sound()
  def sound(self): # Define the required method
    pass   # No actual animal sound is defined here

class Dog(Animal):  # Dog inherites the rule from Animal

  def sound(self):  # Dog provides its version of sound()
    print("Woof!")  # Dog implementation

class Cat(Animal):   # Cat inherites the rule from Animal

  def sound(self):  # Cat provides its version of sound()
    print("Meow!")  # Cat's implementation

dog = Dog()   # Create a Dog object
cat = Cat()   # Create a Cat object

dog.sound()  # Ask Dog to make its sound
cat.sound()  # Ask Cat to makeits sound
```

ABC (Abstract Base Class)  a rule book / blueprint  ,  not a specific object      @abstractmethod = every child class must provide it's own version

```python
from abc import ABC, abstractmethod  # Import abstraction tools

class Shape(ABC): # Create an abstract parent called Shape

  @abstractmethod  # Require every cocncrete shape to implement area()
  def area(self): # Define the common operation : area
    pass # Shape does not choose one formula

class Circle(Shape): # Circle follows the Shape rule

  def area(self): # Circle provides its own area implementation
    print("I calculate the area of circle") # Circle's behaviour

class Square(Shape): # Square follows the shape rule

  def area(self): # Square provides its own area implementation
    print("I calculate the area of square") # Square's behaviour

circle = Circle()   # Create  a Circle object
square = Square() # Create a Square object

circle.area() # Use Circle's implementation
square.area() # Use square's implementation
```

```python
from abc import ABC, abstractmethod  # Import abstract-class tools

class Shape(ABC): # Shape is an abstract base class

  @abstractmethod  # Every cocncrete shape must implement area()
  def area(self): # Define the reuired operation
    pass            # No one formula fits every shape

class Circle(Shape):  # Circle inherots from shape
  def __init__(self, radius):   # Receive the circle's radius
    self.radius = radius        # Circle follows the Shape rule

  def area(self): # Circle's implementation of area()
    return 3.14 * self.radius * self.radius # pi x r x r

class Square(Shape):   # Square inherits from shape
  def __init__(self , side):   # Receive the square's side length
    self.side = side         # Store the side length

  def area(self): # Square's  implementation of area()
    return self.side * self.side   # side x side

circle = Circle(5)   # Create  a Circle with radius 5
square = Square(4) # Create a Square with side 4

print(circle.area()) # Calculate and print Circle's area
print(square.area()) # Calculate and print Square's area
```

threading

thread is a path of work that a program can execute

```python
import threading
import time

def download_file():
    for i in range(1, 4):
        print(f"Download chunk {i}....")
        time.sleep(0.5)

def play_music():
    for i in range(1, 4):
        print(f"Playing beat {i}....")
        time.sleep(0.5)

t1 = threading.Thread(target=download_file)
t2 = threading.Thread(target=play_music)

t1.start()
t2.start()

t1.join()
t2.join()

print("Both tasks are finished!")
```

```python
import threading
import time

def numbers():
    for n in range(1, 6):
        print("Number:", n)
        time.sleep(0.3)

def letters():
    for letter in ["A", "B", "C", "D", "E"]:
        print("Letter:", letter)
        time.sleep(0.3)

t1 = threading.Thread(target=numbers)
t2 = threading.Thread(target=letters)

t1.start()
t2.start()

t1.join()
t2.join()

print("Done!")
```

Multiprocessing

```python
import multiprocessing

def print_num():
    for num in range(5):
        print(num)

def alpha():
    letters = ['a', 'b', 'c', 'd', 'e']

    for letter in letters:
        print(letter)

process1 = multiprocessing.Process(target=print_num)
process2 = multiprocessing.Process(target=alpha)

process1.start()
process1.join()

process2.start()
process2.join()

print("Done!")
```
