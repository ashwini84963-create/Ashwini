Outer and Inner loop with string print statment

Outer Loop = takes one word at a time .
Inner Loop = takes each character from that word

Print each character of a word

```python
words = ["HELLO", "PYTHON"]

for word in words:    # Outer

    for letter in word:     # Inner

        print(letter)
    print("------")
```

Print each word 3 times

```python
words = ["Apple", "Mango", "Banana"]

for word in words:    # Outer
    for i in range(3):  # Inner
        print(word)
    print("-------")
```

Print a string pattern

```python
words = ["A", "AB", "ABC", "ABCD"]

for word in words:
    for letter in word:
        print(letter, end=" ")
    print()
```

Single Inheritance

```python
class Person:
    def speak(self):
        print("Person can speak")

class Student(Person):
    def study(self):
        print("Student can study")

s = Student()
s.speak()
s.study()
```

Multilevel Inheritance

```python
class Vehicle:
    def move(self):
        print("Vehicle can move")

class Car(Vehicle):
    def drive(self):
        print("Car can drive")

class SportsCar(Car, Vehicle):
    def race(self):
        print("Sports car can race")

s = SportsCar()
s.move()
s.drive()
s.race()
```

Abstraction

```python
from abc import ABC ,  abstractmethod

class Car(ABC):
    @abstractmethod
    def start(self):
        pass

class BMW(Car):
    def start(self):
        print("Car starts")

c = BMW()
c.start()
```

Polymorphism

```python
class Cow:
    def sound(self):
        print("Cow says Moo")

class Lion:
    def sound(self):
        print("Lion says Roar")

a = Cow()
b = Lion()

a.sound()
b.sound()
```

Encapsulation

```python
class Bank:
    def __init__(self):
        self.__money = 5000

    def get_money(self):
        return self.__money

    def set_money(self , money):
        self.__money = money

b = Bank()
print(b.get_money())
b.set_money(8000)
print(b.get_money())
```

Daemon thread


a daemon thread is a background thread that runs continously while the main program is running

```python
import threading
import time

def watch():
    while True:
        print(time.strftime("%H:%M:%S"))
        time.sleep(1)

t = threading.Thread(target=watch)
t.daemon = True
t.start()

 # time.sleep(5)
print("Main program ended")
```
