Exception Handling

```python
# setup an error handling block
try:
  print(x_val) # without defining x_val before, we try to print the value of x_val
except:
  print('An exception occured')

# here, since the try block code throws an error because pf x_val not being defined, the accept block is occured
# lets try this without atry block
print(x_val) # this raises a NameError
```

```python
# multiple exception handling
try:
  print(x_val)
except NameError: # exception handling #1
  print("x_val is not defined")
except: # exception handling #2
  print("Something else went wrong")
```

```python
outer_variable = 300

def myfunc():
  print(outer_variable)

myfunc()

print(outer_variable)
```

```python
def func():
  my_salary = 20000

  def myouterfunc():
    print(my_salary)

  myouterfunc()

func()
```

```python
my_salary = 20000
def func():
  print(my_salary)
func()
print(my_salary)
```

Global keyword

if u need to create a global variable, but stuck in thelocal scope, you can use global keyword
the global keyword makes the variable global

```python
def myfunc():
  global global_variable # first u declare it is global variable can be used everywhere
  global_variable = 300
  print(global_variable)

myfunc()

print(global_variable)
```

prog to count number of lines, word, characters

```python
import os, sys
# Import os for file checking and sys for exiting the program

fname = input("enter file name=") # Ask the user to enter the file name
if os.path.isfile(fname): # Check whether the give file exists
  f = open(fname, 'r') # Open the file in read mode
else: # If the file does not exist
  print(fname+"does not exist") # Display a message that the file does not exit
  sys.exit() # Stop the program

cl = cw = cc = 0 # Initialize line count, word count and character count to 0
for line in f: # Read the file one line at a time
  words = line.split() # Split the line into individual words
  cl+= 1 # Increase the line count by 1
  cw+= len(words) # Add the number of words in the line
  cc+= len(line) # Add the number of characters in the line

print("number of lines", cl) # Display the total number of line
print("numner of words", cw) # Display the total number of words
print("number of characters", cc) # Display the total number of characters

f.close() # Close the file
```

```python
import zipfile

# Step 1: create a simple text file to test with
with open("my_note.txt", "w") as f: # will create a new file with given name
  f.write("Hello Class 8! This is a test note.")

# Step 2: Open a new zip file in 'w' (write) node so we get zip file with below name
zip_bag = zipfile.ZipFile("note_backup.zip", "w")

# Step 3: Put our note file inside the zip bag which is variable
zip_bag.write("my_note.txt") # passing name of file which we want to zip

# Step 4: Close the zip bag so it saves properly
zip_bag.close()

# Step 5: Tell the user it worked
print("Success! 'my_note.txt' is now packed inside 'note_backup zip'.")
```

Unzip an entire folder

```python
import zipfile

# Step 1: Open our zip file in 'r' (read) mode
zip_bag = zipfile.ZipFile("note_backup.zip", "r")

# Step 2: Extract (unzip) everything into a new folder named 'my_unzipped_note'
zip_bag.extractall("my_unzipped_note") # so unzip files comes under this folder

# Step 3: Close the zip file
zip_bag.close()

# Step 4: Tell the user it worked
print("Success! File unzipped inside the 'my_unzipped_note' folder.")
```

```python
import os
import shutil

# Step 1: Create a test folder named 'class_photos'
# if same name folder exists then exit which set to True
os.makedirs("class_photos", exist_ok=True)

# Step 2: Put afake picture file inside the folder
# passing the path that inside this folder we have to create this photo file
with open("class_photos/pic1.jpg", "w") as f:
  f.write("Fake image data")

# Step 3: Zip the ENTIRE folder using python's shutil tool
# Format: shutil.make_archive("new_zip_name", "zip", "folder_to_zip")
# It takes 3 arg
shutil.make_archive("photos_backup", "zip", "class_photos")

# Step 4: Tell the user it worked!
print("Sucess! The 'class_photos' folder is now zipped into 'photos_backup.zip.")
```
